```python
np.random.seed(5)
df = pd.DataFrame(
    np.random.randint(1, 100, size=(15, 3)),
    index=[
        # [f'{i}학년' for i in range(1, 4) for _ in range(5)], # 학년
        ['1학년'] * 5 + ['2학년'] * 5 + ['3학년'] * 5,
        # [f'{i}반' for _ in range(3) for i in range(1, 6) ]  # 반
        [f'{i}반' for i in range(1, 6)] * 3
    ],
    columns=['국어', '영어', '과학'],
)
df
```

```python
subject_cols = ['국어', '영어', '과학']
df['합계'] = df[subject_cols].sum(axis=1)
df['평균'] = df['합계'] / len(subject_cols)
df
```

```python
df.index.names = ['학년', '반']
df.groupby('학년').sum()
```

```python
# 4. 나이(Age)를 활용 유아, 10대, 20대, 30대, 40대, 50대, 60대, 노인의 생존자 합을 구해주세요.

def get_group(age):
    if age < 10:
        return '유아'
        
    for i in range(1, 7):
        if age//10 == i :
            return f'{i}0대'
    
    return '노인'
    # if age < 10:
    #     return '유아'
    # elif age < 20:
    #     return '10대'
    # ...
    # elif age < 70:
    #     return '60대'
    # else:
    #     return '노인'

df['Age_group'] = df['Age'].apply(get_group)
df.groupby('Age_group').sum()['Survived']
```
